```python
# TensorFlow program starts like this
import tensorflow as tf 
import numpy as np
import matplotlib.pyplot as plt

sample_size = 50
X_train = np.linspace(0, 1, sample_size)
Y_train = X_train + 0.05 * np.random.normal( size=X_train.size)

plt.figure(dpi=100)
ax = plt.axes()
ax.scatter(X_train, Y_train, label="training data")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Training Data")
plt.show()
```

```python
w = tf.Variable(tf.random.normal(shape=[1], stddev=0.25))
b = tf.Variable(tf.random.normal(shape=[1], stddev=0.25)) 

def predict(x, w, b):
  return w*x + b

def mse(y_pred, y_true):
  return tf.reduce_mean( tf.square(y_pred - y_true)) 

# defferent parameters
epochs = 10
lr = 0.5

for epoch in range(epochs):
  with tf.GradientTape() as tape:
    y_pred = predict( X_train, w, b)
    loss = mse ( y_pred, Y_train)

    gradients = tape.gradient( loss, [w, b])

    # subtract scaled gradients from model parameters
    w.assign_sub( lr * gradients[0])
    b.assign_sub( lr * gradients[1])

    if epoch % 100 == 0:
      print("Epoch: ", epoch, " MSE: ", loss.numpy())

plt.figure(dpi=100)
ax = plt.axes()
ax.scatter(X_train, Y_train, label="training samples")
ax.plot( X_train, predict(X_train, w, b), color="r", label="model fit")
plt.title("Linear Regression: Handcrafted Model")
plt.legend()
plt.show()

    
```
